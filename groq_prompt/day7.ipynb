```python
from groq import Groq
from dotenv import load_dotenv
import os

load_dotenv()

GROQ_API_KEY = os.getenv("GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)


def generate_response(persona_prompt, user_query):
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {
                "role": "system",
                "content": persona_prompt
            },
            {
                "role": "user",
                "content": user_query
            }
        ]
    )

    return response.choices[0].message.content


persona_prompt = """
 You are an expert placement interview assistant.

    Your job is to help students prepare for software engineering
    and AI/ML interviews.
    Give answers in simple and clear English.
    When explaining technical concepts, use simple examples.

    If the user gives an interview answer:
    1. Identify mistakes.
    2. Suggest improvements.
    3. Give a better version of the answer.
"""

user_query = input("You: ")

answer = generate_response(persona_prompt, user_query)

print(answer)
```
